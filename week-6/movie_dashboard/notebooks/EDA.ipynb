```python

# SETUP CELL: IMPORT LIBRARIES AND LOAD DATASET

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from IPython.display import display

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())


# HELPER FUNCTION: PRINT CONCLUSION BELOW EACH GRAPH
def conclusion(message):
    print("\nCONCLUSION:")
    print(message)
    print("-" * 70)


# HELPER FUNCTION: SAFE NUMERIC CORRELATION
def get_corr(col1, col2):
    temp = df[[col1, col2]].dropna()
    if len(temp) < 2:
        return np.nan
    return temp[col1].corr(temp[col2])
```

```python

# VISUALISATION 1: BUDGET HISTOGRAM

data = df["budget"].dropna()

plt.figure(figsize=(9, 5))
sns.histplot(data, bins=35, kde=True)
plt.title("Distribution of Movie Budgets")
plt.xlabel("Movie Budget")
plt.ylabel("Number of Movies")
plt.show()

skew = data.skew()

if skew > 0.5:
    msg = "Movie budgets are positively skewed. A smaller number of movies have very large budgets."
elif skew < -0.5:
    msg = "Movie budgets are negatively skewed."
else:
    msg = "Movie budgets are relatively symmetrical in distribution."

conclusion(msg)
```

```python

# VISUALISATION 3: DURATION HISTOGRAM

data = df["duration"].dropna()

plt.figure(figsize=(9, 5))
sns.histplot(data, bins=30, kde=True)
plt.title("Distribution of Movie Duration")
plt.xlabel("Duration in Minutes")
plt.ylabel("Number of Movies")
plt.show()

conclusion(
    f"The average movie duration is {data.mean():.1f} minutes "
    f"and the median duration is {data.median():.1f} minutes."
)
```

```python

# VISUALISATION 4: BUDGET BOXPLOT

data = df["budget"].dropna()

plt.figure(figsize=(9, 4))
sns.boxplot(x=data)
plt.title("Box Plot of Movie Budgets")
plt.xlabel("Budget")
plt.show()

q1 = data.quantile(0.25)
q3 = data.quantile(0.75)
iqr = q3 - q1
outliers = ((data < q1 - 1.5*iqr) | (data > q3 + 1.5*iqr)).sum()

conclusion(
    f"The median budget is {data.median():,.0f}. "
    f"The IQR method flags {outliers} possible budget outliers. "
    "These should be inspected, not automatically deleted."
)
```

```python

# VISUALISATION 18: COUNTRY VS AVERAGE GROSS

temp = df[["country", "gross"]].dropna()

summary = temp.groupby("country")["gross"].agg(["mean", "count"])
summary = summary[summary["count"] >= 10]
summary = summary.sort_values("mean", ascending=False).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=summary["mean"], y=summary.index)
plt.title("Top 10 Countries by Average Gross Revenue")
plt.xlabel("Average Gross Revenue")
plt.ylabel("Country")
plt.show()

conclusion(
    f"{summary.index[0]} has the highest average gross revenue "
    "among the displayed countries with at least 10 records."
)
```

```python

# VISUALISATION 19: BUDGET GROUP VS GROSS

temp = df[["budget", "gross"]].dropna()
temp = temp[temp["budget"] > 0].copy()

temp["Budget Group"] = pd.qcut(
    temp["budget"],
    q=4,
    labels=["Low", "Medium-Low", "Medium-High", "High"],
    duplicates="drop"
)

summary = temp.groupby("Budget Group", observed=True)["gross"].mean()

plt.figure(figsize=(9, 5))
sns.barplot(x=summary.index, y=summary.values)
plt.title("Average Gross Revenue by Budget Group")
plt.xlabel("Budget Group")
plt.ylabel("Average Gross Revenue")
plt.show()

conclusion(
    f"The {summary.idxmax()} budget group has the highest average gross "
    f"revenue ({summary.max():,.0f})."
)
```

```python

# VISUALISATION 30: GENRE VS COUNTRY HEATMAP

temp = df[["genres", "country", "gross"]].dropna().copy()
temp["genres"] = temp["genres"].str.split("|")
temp = temp.explode("genres")

top_genres = temp["genres"].value_counts().head(8).index
top_countries = temp["country"].value_counts().head(8).index

temp = temp[
    temp["genres"].isin(top_genres) &
    temp["country"].isin(top_countries)
]

pivot = temp.pivot_table(
    index="genres",
    columns="country",
    values="gross",
    aggfunc="mean"
)

plt.figure(figsize=(13, 7))
sns.heatmap(pivot, cmap="YlGnBu", annot=True, fmt=".0f")
plt.title("Average Gross Revenue by Genre and Country")
plt.xlabel("Country")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()

conclusion(
    "The heatmap compares average gross revenue for genre-country "
    "combinations. Darker cells represent higher values in this colour "
    "scale. Some combinations may have very few movies."
)
```
