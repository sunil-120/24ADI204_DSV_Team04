```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, Normalizer
from google.colab import files
```

```python
uploaded = files.upload()

df = pd.read_csv('movie_dataset_encoded.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# Select numerical columns
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Exclude target column
if 'gross' in numerical_columns:
    numerical_columns.remove('gross')

print("Numerical columns:", numerical_columns)
```

```python
# Identify binary columns containing only 0 and 1
binary_columns = [
    col for col in numerical_columns
    if set(df[col].dropna().unique()).issubset({0, 1})
]

# Identify continuous numerical columns
continuous_columns = [
    col for col in numerical_columns
    if col not in binary_columns
]

print("Continuous columns:", continuous_columns)
print("\nBinary columns:", binary_columns)
```

```python
minmax_df = df.copy()

scaler = MinMaxScaler()

minmax_df[continuous_columns] = scaler.fit_transform(
    minmax_df[continuous_columns]
)

print("Min-Max Scaling completed!")
minmax_df.head()
```

```python
minmax_df.to_csv('minmax_scaled_dataset.csv', index=False)

files.download('minmax_scaled_dataset.csv')
```

```python
standard_df = df.copy()

scaler = StandardScaler()

standard_df[continuous_columns] = scaler.fit_transform(
    standard_df[continuous_columns]
)

print("Standardization completed!")
standard_df.head()
```

```python
standard_df.to_csv('standardized_dataset.csv', index=False)

files.download('standardized_dataset.csv')
```

```python
# Check missing values
print("Missing values:")
print(df[continuous_columns].isnull().sum())

# Check infinite values
print("\nInfinite values:")
print(np.isinf(df[continuous_columns]).sum())
```

```python
normalized_df = df.copy()

# Replace infinite values with NaN
normalized_df[continuous_columns] = normalized_df[continuous_columns].replace(
    [np.inf, -np.inf], np.nan
)

# Fill missing values with median
normalized_df[continuous_columns] = normalized_df[continuous_columns].fillna(
    normalized_df[continuous_columns].median()
)

# Apply L2 Normalization
scaler = Normalizer(norm='l2')

normalized_df[continuous_columns] = scaler.fit_transform(
    normalized_df[continuous_columns]
)

print("Normalization completed!")

normalized_df.head()
```

```python
normalized_df.to_csv('normalized_dataset.csv', index=False)

files.download('normalized_dataset.csv')
```
