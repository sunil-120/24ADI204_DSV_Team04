```python
import pandas as pd
from google.colab import files

uploaded = files.upload()

df = pd.read_csv('feature_engineered_dataset.csv')

print("Original shape:", df.shape)
df.head()
```

```python
# Encoding duration category
duration_mapping = {
    'Unknown': 0,
    'Short': 1,
    'Medium': 2,
    'Long': 3
}

df['duration_category'] = df['duration_category'].map(duration_mapping)


# Encoding budget category
budget_mapping = {
    'Unknown': 0,
    'Low': 1,
    'Medium': 2,
    'High': 3
}

df['budget_category'] = df['budget_category'].map(budget_mapping)


# Encoding release decade
decade_mapping = {
    'Unknown': 0,
    '1920s': 1, '1930s': 2, '1940s': 3,
    '1950s': 4, '1960s': 5, '1970s': 6,
    '1980s': 7, '1990s': 8, '2000s': 9,
    '2010s': 10, '2020s': 11
}

df['release_decade'] = df['release_decade'].map(decade_mapping)

print("Ordinal encoding completed.")
```

```python
# Select categorical columns for one-hot encoding
categorical_columns = [
    'content_rating',
    'language',
    'country',
    'color'
]

# Handle missing values
for col in categorical_columns:
    df[col] = df[col].fillna('Unknown')

# Apply one-hot encoding
df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

print("One-hot encoding completed.")
```

```python
print("Shape after encoding:", df.shape)

print("\nRemaining categorical columns:")
print(df.select_dtypes(include=['object']).columns.tolist())

df.head()
```

```python
df.to_csv('movie_dataset_encoded.csv', index=False)

files.download('movie_dataset_encoded.csv')
```
