```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Dataset Shape:", df.shape)

display(df.head())
```

```python
print("Existing Columns:")

print(df.columns.tolist())

print("\nDataset Information:")

df.info()
```

```python
df['release_decade'] = (df['title_year'] // 10 * 10).astype('Int64').astype(str) + 's'

print(df[['title_year', 'release_decade']].head())
```

```python
df['number_of_genres'] = df['genres'].apply(
    lambda x: len(str(x).split('|')) if pd.notna(x) else 0
)

print(df[['genres', 'number_of_genres']].head())
```

```python
def duration_category(duration):

    if pd.isna(duration):
        return 'Unknown'

    elif duration < 90:
        return 'Short'

    elif duration <= 120:
        return 'Medium'

    else:
        return 'Long'


df['duration_category'] = df['duration'].apply(duration_category)

print(df[['duration', 'duration_category']].head())
```

```python
Q2 = df['budget'].median()

Q3 = df['budget'].quantile(0.75)


def budget_category(budget):

    if pd.isna(budget):
        return 'Unknown'

    elif budget <= Q2:
        return 'Low'

    elif budget <= Q3:
        return 'Medium'

    else:
        return 'High'


df['budget_category'] = df['budget'].apply(budget_category)

print(df[['budget', 'budget_category']].head())
```

```python
df['number_of_plot_keywords'] = df['plot_keywords'].apply(
    lambda x: len(str(x).split('|')) if pd.notna(x) else 0
)

print(df[['plot_keywords', 'number_of_plot_keywords']].head())
```

```python
actor_columns = [
    'actor_1_facebook_likes',
    'actor_2_facebook_likes',
    'actor_3_facebook_likes'
]

df['cast_popularity'] = df[actor_columns].fillna(0).sum(axis=1)

print(df[actor_columns + ['cast_popularity']].head())
```

```python
df['movie_age'] = 2026 - df['title_year']

print(df[['title_year', 'movie_age']].head())
```

```python
df['production_country_count'] = df['country'].apply(
    lambda x: len(str(x).split('|')) if pd.notna(x) else 0
)

print(df[['country', 'production_country_count']].head())
```

```python
new_features = [
    'release_decade',
    'number_of_genres',
    'duration_category',
    'budget_category',
    'number_of_plot_keywords',
    'cast_popularity',
    'movie_age',
    'production_country_count'
]

display(df[new_features].head(10))
```

```python
print("Updated Dataset Shape:", df.shape)

print("\nTotal Number of Features:", len(df.columns))

print("\nNewly Created Features:", len(new_features))

print("\nNew Features:")

print(new_features)
```

```python
output_file = "feature_engineered_dataset.csv"

df.to_csv(output_file, index=False)

print("Feature Engineering Completed Successfully!")

files.download(output_file)
```
