```python
# ==========================================
# WEEK 2 - DATA CLEANING
# MOVIE SUCCESS PREDICTOR
# ==========================================


# 1. Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files


# 2. Load Original Dataset

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Original Dataset Shape:", df.shape)

display(df.head())


# 3. Remove Unwanted Columns

columns_to_remove = [
    'movie_imdb_link',
    'imdb_score',
    'num_voted_users',
    'num_user_for_reviews',
    'num_critic_for_reviews'
]

df.drop(columns=columns_to_remove, inplace=True)

print("\nColumns Removed Successfully!")
print("Updated Dataset Shape:", df.shape)


# 4. Check Missing Values

print("\nMissing Values in Each Column:")

missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    'Missing Values': missing_values,
    'Percentage': (missing_values / len(df)) * 100
})

display(missing_report[missing_report['Missing Values'] > 0])


# 5. Visualize Missing Values

plt.figure(figsize=(14, 6))

sns.heatmap(df.isnull(), cbar=False, cmap='viridis')

plt.title("Missing Values in Movie Dataset")
plt.xlabel("Columns")
plt.ylabel("Movies")

plt.show()


# 6. Handle Missing Values in Numerical Columns

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:

    if col != 'gross':
        df[col] = df[col].fillna(df[col].median())

print("\nMissing Values in Numerical Columns Handled Using Median.")


# 7. Handle Missing Values in Categorical Columns

categorical_columns = df.select_dtypes(include='object').columns

for col in categorical_columns:

    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing Values in Categorical Columns Handled Using Mode.")


# 8. Check Remaining Missing Values

print("\nRemaining Missing Values:")

print(df.isnull().sum()[df.isnull().sum() > 0])


# 9. Visualize Outliers Using Boxplots

columns = ['budget', 'gross', 'duration']

plt.figure(figsize=(15, 5))

for i, col in enumerate(columns, 1):

    plt.subplot(1, 3, i)

    sns.boxplot(y=df[col])

    plt.title(f'Boxplot of {col}')

plt.tight_layout()

plt.show()


# 10. Detect Outliers Using IQR Method

print("\n========== IQR OUTLIER DETECTION ==========")

for col in columns:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print("\nAttribute:", col)

    print("Q1:", Q1)

    print("Q3:", Q3)

    print("IQR:", IQR)

    print("Lower Bound:", lower_bound)

    print("Upper Bound:", upper_bound)

    print("Number of Outliers:", len(outliers))

print("\nOutliers Detected but Not Removed.")


# 11. Check Duplicate Records

duplicates = df.duplicated().sum()

print("\nNumber of Duplicate Records:", duplicates)

# Duplicates are identified but not removed.


# 12. Generate Final Data Quality Report

print("\n========== FINAL DATA QUALITY REPORT ==========")

print("\nDataset Shape:", df.shape)

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nRemaining Missing Values:")

print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nDuplicate Records:", df.duplicated().sum())

print("\nData Types:")

print(df.dtypes)

print("\nStatistical Summary:")

display(df.describe())


# 13. Save Cleaned Dataset

output_file = "movie_dataset_cleaned.csv"

df.to_csv(output_file, index=False)

print("\nCleaned Dataset Saved Successfully!")

files.download(output_file)
```
