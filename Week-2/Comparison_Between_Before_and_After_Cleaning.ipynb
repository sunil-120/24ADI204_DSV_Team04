```python
# COMMAND 1: Import Required Libraries

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# COMMAND 2: Load Before and After Cleaning Datasets

before = pd.read_csv('/content/movie_metadata.csv')
after = pd.read_csv('/content/imdb_5000_cleaned.csv')

print("Both datasets loaded successfully!")
```

```python
# COMMAND 3: Compare Dataset Size

print("BEFORE CLEANING")
print("Rows    :", before.shape[0])
print("Columns :", before.shape[1])

print("\nAFTER CLEANING")
print("Rows    :", after.shape[0])
print("Columns :", after.shape[1])
```

```python
# COMMAND 4: View Original Dataset

print("First 5 Rows - BEFORE CLEANING")
display(before.head())
```

```python
# COMMAND 5: View Cleaned Dataset

print("First 5 Rows - AFTER CLEANING")
display(after.head())
```

```python
# COMMAND 6: Compare Missing Values

missing_before = before.isnull().sum().sum()
missing_after = after.isnull().sum().sum()

print("Missing Values BEFORE Cleaning :", missing_before)
print("Missing Values AFTER Cleaning  :", missing_after)
```

```python
# COMMAND 7: Compare Duplicate Rows

duplicates_before = before.duplicated().sum()
duplicates_after = after.duplicated().sum()

print("Duplicate Rows BEFORE Cleaning :", duplicates_before)
print("Duplicate Rows AFTER Cleaning  :", duplicates_after)
```

```python
# COMMAND 8: Create Cleaning Summary Table

comparison = pd.DataFrame({
    'Before Cleaning': [
        before.shape[0],
        before.shape[1],
        missing_before,
        duplicates_before
    ],

    'After Cleaning': [
        after.shape[0],
        after.shape[1],
        missing_after,
        duplicates_after
    ]

}, index=[
    'Number of Rows',
    'Number of Columns',
    'Missing Values',
    'Duplicate Rows'
])

display(comparison)
```

```python
# COMMAND 9: Visualize Missing Values Before vs After

plt.figure(figsize=(8,5))

plt.bar(
    ['Before Cleaning', 'After Cleaning'],
    [missing_before, missing_after]
)

plt.title('Missing Values Before vs After Cleaning')
plt.xlabel('Dataset')
plt.ylabel('Number of Missing Values')

plt.tight_layout()
plt.show()
```

```python
# COMMAND 10: Visualize Duplicate Rows Before vs After

plt.figure(figsize=(8,5))

plt.bar(
    ['Before Cleaning', 'After Cleaning'],
    [duplicates_before, duplicates_after]
)

plt.title('Duplicate Rows Before vs After Cleaning')
plt.xlabel('Dataset')
plt.ylabel('Number of Duplicate Rows')

plt.tight_layout()
plt.show()
```

```python
# COMMAND 11: Visualize Number of Rows

plt.figure(figsize=(8,5))

plt.bar(
    ['Before Cleaning', 'After Cleaning'],
    [before.shape[0], after.shape[0]]
)

plt.title('Number of Rows Before vs After Cleaning')
plt.xlabel('Dataset')
plt.ylabel('Number of Rows')

plt.tight_layout()
plt.show()
```

```python
# COMMAND 12: Missing Values by Column

missing_before_column = before.isnull().sum()
missing_after_column = after.isnull().sum()

columns_with_missing = missing_before_column[
    missing_before_column > 0
].index

missing_comparison = pd.DataFrame({

    'Before Cleaning':
        missing_before_column[columns_with_missing],

    'After Cleaning':
        missing_after_column[columns_with_missing]

})

display(missing_comparison)
```

```python
# COMMAND 13: Visualize Missing Values by Column

plt.figure(figsize=(12,6))

x = np.arange(len(columns_with_missing))
width = 0.35

plt.bar(
    x - width/2,
    missing_comparison['Before Cleaning'],
    width,
    label='Before Cleaning'
)

plt.bar(
    x + width/2,
    missing_comparison['After Cleaning'],
    width,
    label='After Cleaning'
)

plt.title('Missing Values by Column - Before vs After Cleaning')

plt.xlabel('Columns')
plt.ylabel('Number of Missing Values')

plt.xticks(
    x,
    columns_with_missing,
    rotation=90
)

plt.legend()

plt.tight_layout()
plt.show()
```

```python
# COMMAND 14: Select Important Movie Variables

features = [
    'budget',
    'gross',
    'imdb_score',
    'duration'
]

features = [
    feature
    for feature in features
    if feature in before.columns and feature in after.columns
]

print("Features selected for comparison:")
print(features)
```

```python
# COMMAND 15: Compare Distributions Using Boxplots

for feature in features:

    plt.figure(figsize=(8,5))

    plt.boxplot(
        [
            before[feature].dropna(),
            after[feature].dropna()
        ],
        labels=[
            'Before Cleaning',
            'After Cleaning'
        ]
    )

    plt.title(
        'Before vs After Cleaning - ' + feature
    )

    plt.ylabel(feature)

    plt.tight_layout()
    plt.show()
```

```python
# COMMAND 16: Final Cleaning Summary

print("=" * 60)
print("FINAL DATA CLEANING SUMMARY")
print("=" * 60)

print("\nOriginal Dataset:")
print("Rows    :", before.shape[0])
print("Columns :", before.shape[1])

print("\nCleaned Dataset:")
print("Rows    :", after.shape[0])
print("Columns :", after.shape[1])

print("\nCleaning Changes:")
print("Rows Removed        :", before.shape[0] - after.shape[0])
print("Missing Values Fixed:", missing_before - missing_after)
print("Duplicates Removed  :", duplicates_before - duplicates_after)

print("\nCleaning comparison completed successfully!")
```
